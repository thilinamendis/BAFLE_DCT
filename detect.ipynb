### Setup Paths and Preprocess Function for GNCNN

```python
import torch
import torchvision.transforms as transforms
from PIL import Image

# Define preprocessing transform for GNCNN
gncnn_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.Grayscale(num_output_channels=1),
    transforms.ToTensor(),  # Converts to [0,1] float and shape [C,H,W]
    transforms.Normalize((0.5,), (0.5,))  # Normalize to [-1,1]
])

def preprocess_for_gncnn(np_image):
    """
    Preprocess an image (numpy) for GNCNN input.
    Args:
        np_image (numpy array): H x W x 3 RGB uint8
    Returns:
        torch.Tensor: preprocessed image, shape [1, 1, 256, 256]
    """
    pil_image = Image.fromarray(np_image)
    tensor_image = gncnn_transform(pil_image)
    tensor_image = tensor_image.unsqueeze(0)  # Add batch dimension
    return tensor_image
```
